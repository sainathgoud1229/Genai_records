### STEP 2: PREPARE SAMPLE DOCUMENTS

Create a collection of sample documents related to artificial intelligence, machine learning, Python, deep learning, and natural language processing.

These documents will be converted into embeddings and used for semantic search.

```python
documents = [
    "Artificial Intelligence enables machines to perform tasks that require human intelligence.",
    "Machine Learning is a subset of AI that learns patterns from data.",
    "Deep Learning uses neural networks with multiple layers.",
    "Python is a popular programming language for AI and machine learning.",
    "Natural Language Processing helps computers understand human language.",
    "Computer Vision enables machines to analyze and understand images and videos.",
    "Generative AI can create text, images, music, and other content.",
    "Data Science involves collecting, analyzing, and interpreting data."
]

print("Total documents:", len(documents))

for i, doc in enumerate(documents, start=1):
    print(f"{i}. {doc}")
```

###  LOAD PRE-TRAINED EMBEDDING MODEL

Load the all-MiniLM-L6-v2 sentence-transformer model.

This model converts sentences into numerical vectors that capture their semantic meaning.

```python
import numpy as np
from sentence_transformers import SentenceTransformer

print("Libraries imported successfully!")
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")

print("Embedding model loaded successfully!")
```

```python
documents = [
    "Artificial Intelligence enables machines to perform tasks that require human intelligence.",
    "Machine Learning is a subset of AI that learns patterns from data.",
    "Deep Learning uses neural networks with multiple layers.",
    "Python is a popular programming language for AI and machine learning.",
    "Natural Language Processing helps computers understand human language.",
    "Computer Vision enables machines to analyze and understand images and videos.",
    "Generative AI can create text, images, music, and other content.",
    "Data Science involves collecting, analyzing, and interpreting data."
]

print("Documents created successfully!")
print("Total documents:", len(documents))

for i, doc in enumerate(documents, start=1):
    print(f"{i}. {doc}")
```

```python
embeddings = model.encode(
    documents,
    convert_to_numpy=True,
    normalize_embeddings=True
)

print("Embeddings generated successfully!")
print("Embedding shape:", embeddings.shape)
print("First document embedding (first 10 values):")
print(embeddings[0][:10])
```

## CREATE VECTOR STORE

Store the generated document embeddings in a NumPy array.
This vector store will be used to perform semantic similarity searches.

```python
vector_store = embeddings

print("Vectors stored successfully!")
print("Number of vectors:", vector_store.shape[0])
print("Vector dimensions:", vector_store.shape[1])
```

##  PERFORM SEMANTIC SEARCH

Enter a search query and convert it into an embedding.
Calculate cosine similarity between the query embedding and stored document embeddings.
Retrieve and display the top 3 most relevant documents.

```python
query = "How do computers understand human language?"

query_embedding = model.encode(
    [query],
    convert_to_numpy=True,
    normalize_embeddings=True
)

similarities = np.dot(vector_store, query_embedding[0])

top_indices = np.argsort(similarities)[::-1][:3]

print("Search Query:", query)
print("\nTop 3 Relevant Documents:\n")

for rank, idx in enumerate(top_indices, start=1):
    print(f"{rank}. {documents[idx]}")
    print(f"   Similarity Score: {similarities[idx]:.4f}\n")
```

## RESULT

Text embeddings were successfully generated using the Sentence Transformers model (all-MiniLM-L6-v2). The embeddings were stored in a NumPy-based vector store, and semantic search was performed using cosine similarity to retrieve the top 3 relevant documents based on the query.

FAISS was not used because Windows Application Control blocked its native library import. NumPy was used as an alternative for vector storage and similarity search.

# By using langchain with huggingface for Faiss

```python
from langchain_core.documents import Document
from langchain_huggingface import HuggingFaceEmbeddings
import numpy as np
```

```python

documents = [
    Document(page_content="AI helps machines learn from data."),
    Document(page_content="Python is a popular programming language."),
    Document(page_content="Deep learning uses neural networks."),
    Document(page_content="Plants need sunlight and water to grow.")
]

# Step 2: Load embedding model
embeddings_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

texts = [doc.page_content for doc in documents]

document_vectors = embeddings_model.embed_documents(texts)

vector_store = np.array(document_vectors, dtype=np.float32)

# Step 4: Perform semantic search
query = "How do machines learn?"

query_vector = np.array(
    embeddings_model.embed_query(query),
    dtype=np.float32
)

# Step 5: Calculate cosine similarity
document_norms = np.linalg.norm(vector_store, axis=1)
query_norm = np.linalg.norm(query_vector)

similarities = np.dot(vector_store, query_vector) / (
    document_norms * query_norm
)

# Step 6: Retrieve top 2 results
top_indices = np.argsort(similarities)[::-1][:2]

# Step 7: Display results
print("Query:", query)
print("\nSearch Results:")

for idx in top_indices:
    print("-", documents[idx].page_content)
    print("Similarity Score:", round(float(similarities[idx]), 4))
```
