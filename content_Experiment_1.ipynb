#### AIM  : To generate content using Generative AI and reusable prompt templates.

### Algorithm :
1. Start the program. 
2. Define a prompt template for content generation. 
3. Insert the topic into the template. 
4. Generate content using the template. 
5. Display the generated content. 
6. Stop the program.


```python
# CODE:

from string import Template

# Define a reusable prompt template
prompt = Template(
    "Write a short paragraph about $topic "
    "in simple English with a suitable title."
)

# Insert the topic into the template
topic = "Artificial Intelligence"
final_prompt = prompt.substitute(topic=topic)

print("PROMPT:")
print(final_prompt)

print("\nGENERATED CONTENT:")
print(
    "Artificial Intelligence is a technology that enables "
    "machines to perform tasks that normally require human "
    "intelligence. It is used in healthcare, education, "
    "transportation, and many other fields."
)
```
