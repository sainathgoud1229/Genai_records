### Aim 
- To develop a customer support assistant using LangChain, an LLM, a tool, and conversational memory.

### Algorithm


1. Import LangChain components. 
2. Define a customer support tool. 
3. Create a prompt template. 
4. Connect the prompt to a language model. 
5. Store the conversation history. 
6. Send a customer query to the assistant. 
7. Display the response.
