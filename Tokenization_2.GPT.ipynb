
## Tokenization and Text Generation with GPT-2

### Aim
To tokenize input text and generate new text using the GPT-2 language model.

### Objective
- Understand tokenization in Large Language Models.
- Load a pretrained GPT-2 model.
- Convert text into tokens and numerical token IDs.
- Generate text using GPT-2.
- Decode generated token IDs into readable text.

# Algorithms 

1.Import the Transformers library.



2.Load the GPT-2 tokenizer and pretrained model.



3.Provide an input prompt.



4.Tokenize the prompt into token IDs.



5.Generate text using GPT-2.



6.Decode the generated token IDs.



7.Display the output.

```python
import sys

print(sys.executable)
```

```python
import transformers
import torch

print("Transformers:", transformers.__version__)
print("PyTorch:", torch.__version__)
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer

print("GPT-2 libraries imported successfully!")
```

```python
tokenizer = GPT2Tokenizer.from_pretrained("gpt2")

print("GPT-2 tokenizer loaded successfully!")
```

```python
model = GPT2LMHeadModel.from_pretrained("gpt2")

print("GPT-2 model loaded successfully!")
```

```python
prompt = "Artificial intelligence is"

print("Prompt:", prompt)
```

```python
tokens = tokenizer.encode(prompt, return_tensors="pt")

print("Tokens:", tokenizer.tokenize(prompt))
print("Token IDs:", tokens.tolist())
```

```python
output = model.generate(
    tokens,
    max_new_tokens=30,
    do_sample=False,
    pad_token_id=tokenizer.eos_token_id
)

print("Text generation completed!")
```

```python
generated_text = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)

print("\nGenerated Text:")
print(generated_text)
```

```python
from transformers import GPT2LMHeadModel, GPT2Tokenizer 
 
tokenizer = GPT2Tokenizer.from_pretrained("gpt2") 
model = GPT2LMHeadModel.from_pretrained("gpt2") 
 
prompt = "Artificial intelligence is" 
 
tokens = tokenizer.encode(prompt, return_tensors="pt") 
 
print("Tokens:", tokenizer.tokenize(prompt)) 
print("Token IDs:", tokens.tolist()) 
 
output = model.generate( 

 
    tokens, 
    max_new_tokens=30, 
    do_sample=False, 
    pad_token_id=tokenizer.eos_token_id 
) 
 
generated_text = tokenizer.decode( 
    output[0], skip_special_tokens=True 
) 
 
print("\nGenerated Text:") 
print(generated_text)
```
