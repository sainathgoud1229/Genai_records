
## Misinformation and Phishing Detection

### Aim

To detect potentially misleading information and phishing messages using a simple text classification program.

### Algorithm

1. Import the required Python libraries.
2. Define sample messages.
3. Create keywords associated with suspicious links and misleading claims.
4. Preprocess each message into lowercase text.
5. Check for suspicious phrases and URL patterns.
6. Classify each message based on the detected indicators.
7. Display the message, classification, and result.

#### Libraries

```python
import re


def detect_message(text):

    text_lower = text.lower()

    phishing_keywords = [
        "verify your account",
        "password immediately",
        "claim your prize",
        "urgent action required",
        "send your otp"
    ]

    misinformation_keywords = [
        "guaranteed cure",
        "100% proven false",
        "share before it is deleted",
        "secret information they hide"
    ]

    suspicious_url = bool(
        re.search(r"https?://", text_lower)
    )

    phishing_score = sum(
        keyword in text_lower
        for keyword in phishing_keywords
    )

    misinformation_score = sum(
        keyword in text_lower
        for keyword in misinformation_keywords
    )

    if phishing_score > 0:
        label = "Potential Phishing"

    elif misinformation_score > 0:
        label = "Potential Misinformation"

    elif suspicious_url:
        label = "Review Link Before Opening"

    else:
        label = "No Listed Warning Indicators"

    return label
```

```python
messages = [

    "Verify your account and send your OTP immediately.",

    "This secret information they hide is a guaranteed cure.",

    "Visit https://example.com to read the article.",

    "The library opens at 9 AM."

]

print("Sample messages defined successfully!")
print("Total messages:", len(messages))
```

### Detect and Classify Messages

```python
for message in messages:
    
    print("Message:", message)

    print("Classification:", detect_message(message))

    print()
```

### Output

The program successfully classified four sample messages.

1. OTP verification message – Potential Phishing.
2. Misleading health claim – Potential Phishing.
3. Message containing a URL – Review Link Before Opening.
4. Normal library announcement – No Listed Warning Indicators.

### Result

A rule-based text classifier was developed to flag potential phishing messages and suspicious content. The second sample is classified as phishing because the current code checks for phishing indicators first; changing the priority or using multi-label classification would identify both categories. The system is a basic demonstration and cannot reliably establish whether a claim is true or false.
