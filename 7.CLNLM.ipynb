### Train a Character-Level Neural Language Model.



## Aim

To train a character-level neural language model that learns character patterns and generates new text.

## Algorithm

1. Import PyTorch and required libraries.
2. Prepare a training text corpus.
3. Create a vocabulary of unique characters.
4. Encode characters into numerical IDs.
5. Prepare input and target sequences.
6. Define a neural network.
7. Train the model using a loss function and optimizer.
8. Generate text by predicting subsequent characters.
9. Display the generated text.

## Libraries


```python
import os

os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import torch
import torch.nn as nn

print("PyTorch imported successfully!")
print("PyTorch Version:", torch.__version__)
print("Test Tensor:", torch.tensor([1, 2, 3]))
```

```python
torch.manual_seed(42)

text = "hello world " * 100

print("Training text prepared successfully!")
print("Sample text:", text[:50])
print("Total characters:", len(text))
```

```python
chars = sorted(set(text))
stoi = {ch: i for i, ch in enumerate(chars)}
itos = {i: ch for ch, i in stoi.items()}

print("Unique characters:", chars)
print("Character to ID:", stoi)
print("ID to character:", itos)
print("Vocabulary size:", len(chars))
```

```python
data = torch.tensor(
    [stoi[ch] for ch in text], dtype=torch.long
)

print("Encoded data:", data)
print("Data shape:", data.shape)
```

```python
X = data[:-1].unsqueeze(1)
Y = data[1:]

print("Input shape:", X.shape)
print("Target shape:", Y.shape)
```

```python
class CharModel(nn.Module):
    
    def __init__(self, vocab_size):

        super().__init__()

        self.embedding = nn.Embedding(vocab_size, 16)

        self.rnn = nn.GRU(16, 32, batch_first=True)

        self.fc = nn.Linear(32, vocab_size)

    def forward(self, x):

        x = self.embedding(x)

        x, _ = self.rnn(x)

        return self.fc(x)
```

```python
model = CharModel(len(chars))

optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

loss_fn = nn.CrossEntropyLoss()

print("Character-Level GRU Model Created Successfully!")
print(model)
```

```python
for epoch in range(100):
    
    optimizer.zero_grad()

    logits = model(X).squeeze(1)

    loss = loss_fn(logits, Y)

    loss.backward()

    optimizer.step()

print("Training Loss:", round(loss.item(), 4))
```
