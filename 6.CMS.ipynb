### Combined Memory System



## Aim

To maintain conversational history and a running summary of a conversation using a combined memory system.

# Algorithm
1. Import the memory and language model libraries.
2. Create a conversation buffer memory.
3. Create a conversation summary memory.
4. Combine the memory components.
5. Save sample conversation messages.
6. Load and display the combined memory

#### Libraries

```python
import os

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_classic.memory import (
    ConversationBufferMemory,
    ConversationSummaryMemory,
    CombinedMemory
)

print("Libraries imported successfully!")
```

```python
import os
from getpass import getpass

os.environ["GOOGLE_API_KEY"] = getpass("Enter Google API Key: ")
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

print("Gemini LLM initialized successfully!")
```

```python
buffer_memory = ConversationBufferMemory(
    memory_key="chat_history",
    return_messages=True
)

print("Conversation buffer memory created!")
```

```python
summary_memory = ConversationSummaryMemory(
    llm=llm,
    memory_key="conversation_summary"
)

print("Conversation summary memory created!")
```

```python
combined_memory = CombinedMemory(
    memories=[buffer_memory, summary_memory]
)

print("Combined memory system created successfully!")
```

```python
combined_memory.save_context(
    {"input": "My name is LUCKY."},
    {"output": "Hello LUCKY!"}
)

combined_memory.save_context(
    {"input": "I am learning Generative AI."},
    {"output": "That is a useful skill."}
)

print("Conversation messages saved successfully!")
```

```python
print("Combined Memory:")
print(combined_memory.load_memory_variables({}))
```

## Result

A combined memory system was demonstrated using conversation buffer memory and conversation summary memory. The summary was generated using the Gemini language model. The system maintained conversational history and generated a running summary of the conversation.
