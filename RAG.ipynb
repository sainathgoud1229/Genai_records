##       Three types of retrievers in LangChain 


### AIM

To implement similarity search, maximum marginal relevance (MMR), and multi-query retrieval in LangChain.

## ALGORITHM

1. Create a collection of documents.
2. Generate embeddings for each document.
3. Store the embeddings in a FAISS vector store.
4. Configure a similarity retriever.
5. Configure an MMR retriever.
6. Configure a multi-query retriever using an LLM.
7. Submit a query and display retrieved documents.

```python
import os
from langchain_core.documents import Document
from langchain_openai import ChatOpenAI
from langchain_classic.retrievers import MultiQueryRetriever
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

print("All libraries imported successfully!")
```

```python
import os
from getpass import getpass

os.environ["OPENAI_API_KEY"] = getpass("Enter OpenAI API Key: ")
```

```python
docs = [
    Document(page_content="AI is used in medical diagnosis."),
    Document(page_content="Machine learning learns from examples."),
    Document(page_content="Neural networks power deep learning."),
    Document(page_content="AI supports automated customer service."),
    Document(page_content="Plants require light for photosynthesis.")
]

print("Documents created successfully!")
print("Total documents:", len(docs))
```

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embeddings model loaded successfully!")
```

```python
db = FAISS.from_documents(docs, embeddings)

print("FAISS vector store created successfully!")
```

```python
similarity = db.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 2}
)

print("Similarity retriever configured!")
```

```python
mmr = db.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 2, "fetch_k": 4}
)

print("MMR retriever configured!")
```

```python
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0)

print("LLM configured successfully!")
```

```python
multi_query = MultiQueryRetriever.from_llm(
    retriever=similarity,
    llm=llm
)

print("Multi-query retriever configured!")
```

```python
query = "Applications of artificial intelligence"

for name, retriever in [
    ("Similarity", similarity),
    ("MMR", mmr),
    ("Multi-query", multi_query)
]:
    print(f"\n{name} Retriever:")
    results = retriever.invoke(query)
    for doc in results:
        print("-", doc.page_content)
```

### Result

Three retrieval strategies — Similarity Search, Maximum Marginal Relevance (MMR), and Multi-Query Retrieval — were implemented using LangChain. The retrievers were configured and tested with a sample query to retrieve relevant documents from the vector store.
