
###  Automated Exploratory Data Analysis (EDA) with Generative AI

### Aim

To perform automated exploratory data analysis and generate a natural-language summary of a dataset using an LLM.



### Algorithm

1. Import Pandas and the LLM library.
2. Load a dataset.
3. Display its dimensions and data types.
4. Calculate missing values and descriptive statistics.
5. Prepare an EDA summary.
6. Send the summary to the LLM for interpretation.
7. Display the generated insights.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [21, 25, 30, 28, 35],
    "Salary": [25000, 35000, 50000, 42000, 60000],
    "Department": [
        "IT", "HR", "IT", "Finance", "IT"
    ]
})

print("Dataset:")
print(df)
```

```python
print("Dataset Shape:")
print(df.shape)

print("\nData Types:")
print(df.dtypes)
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Descriptive Statistics:")
print(df.describe(include="all"))
```

```python
summary = f"""
Dataset shape:
{df.shape}

Data types:
{df.dtypes.to_string()}

Missing values:
{df.isnull().sum().to_string()}

Statistics:
{df.describe(include='all').to_string()}
"""

print("EDA Summary:")
print(summary)
```

##### Generate AI-Powered EDA Insights Using Gemini

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0
)

prompt = (
    "Analyze this dataset summary. Identify patterns, "
    "missing values, and limitations. Do not invent facts.\n"
    + summary
)

response = llm.invoke(prompt)

print("AI-Generated EDA Insights:")
print(response.content)
```

### Output

The dataset was successfully loaded and analyzed using Pandas. Dataset dimensions, data types, missing values, and descriptive statistics were displayed.

Gemini generated natural-language insights identifying dataset patterns, missing values, and limitations.
